```python
# Create a backup of the original dataset
df_original = df.copy()

# Remove duplicate rows
df = df.drop_duplicates()

# Remove rows with missing values
df = df.dropna()

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Display cleaning results
print("Data cleaning completed!")
print("Original dataset shape:", df_original.shape)
print("Cleaned dataset shape:", df.shape)
print("Remaining missing values:", df.isnull().sum().sum())
print("Remaining duplicate rows:", df.duplicated().sum())
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], errors='coerce')

print("Date conversion completed!")
print(df[['Order Date', 'Ship Date']].dtypes)
```

```python
df['Order Year'] = df['Order Date'].dt.year
df['Order Month'] = df['Order Date'].dt.month
df['Order Month Name'] = df['Order Date'].dt.month_name()
df['Shipping Days'] = (df['Ship Date'] - df['Order Date']).dt.days

print("New columns created!")
df[['Order Date', 'Order Year', 'Order Month Name', 'Shipping Days']].head()
```

```python
print("Total Sales:", round(df['Sales'].sum(), 2))
print("Total Profit:", round(df['Profit'].sum(), 2))
print("Total Orders:", df['Order ID'].nunique())
print("Total Customers:", df['Customer ID'].nunique())
print("Total Products:", df['Product ID'].nunique())
print("Average Order Sales:", round(df['Sales'].mean(), 2))
```

```python

# Select useful features for predicting sales
features = [
    'Ship Mode',
    'Segment',
    'Region',
    'Category',
    'Sub-Category',
    'Quantity',
    'Discount',
    'Shipping Days',
    'Order Month'
]

target = 'Sales'

# Create the model dataset
model_df = df[features + [target]].copy()

# Remove missing and invalid values
model_df = model_df.dropna()
model_df = model_df[model_df['Shipping Days'] >= 0]
model_df = model_df[model_df['Sales'] >= 0]

print("Machine learning dataset prepared!")
print("Dataset shape:", model_df.shape)
print("\nFirst five rows:")
display(model_df.head())
print("\nMissing values:", model_df.isnull().sum().sum())
```

```python

X = model_df[features]
y = model_df[target]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("\nTarget variable: Sales")
```

```python

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
```

```python

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor

categorical_features = [
    'Ship Mode', 'Segment', 'Region',
    'Category', 'Sub-Category'
]

numeric_features = [
    'Quantity', 'Discount',
    'Shipping Days', 'Order Month'
]

preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore'),
         categorical_features),
        ('num', 'passthrough', numeric_features)
    ]
)

model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('regressor', RandomForestRegressor(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    ))
])

model.fit(X_train, y_train)

print("Random Forest model trained successfully!")
```

```python

from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

y_pred = model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("MODEL EVALUATION RESULTS")
print("------------------------")
print(f"Mean Absolute Error (MAE): ${mae:.2f}")
print(f"Root Mean Squared Error (RMSE): ${rmse:.2f}")
print(f"R² Score: {r2:.4f}")
```

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))
plt.scatter(y_test, y_pred, alpha=0.5)
plt.xlabel("Actual Sales")
plt.ylabel("Predicted Sales")
plt.title("Actual vs Predicted Retail Sales")

limits = [
    min(y_test.min(), y_pred.min()),
    max(y_test.max(), y_pred.max())
]
plt.plot(limits, limits, linestyle='--')
plt.tight_layout()
plt.show()
```
